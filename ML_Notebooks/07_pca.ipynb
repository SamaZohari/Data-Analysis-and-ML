```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA
```

```python
# Generate a sine scatter
n_points = 200
min_x_range = 2
max_x_range = 6

# X_base = np.random.uniform(low=min_x_range, high=max_x_range, size=n_points)
X_base = np.linspace(start=min_x_range, stop=max_x_range, num=n_points)
noise = np.random.normal(loc=0, scale=0.5, size=n_points)

Y_base = X_base + noise

ax = plt.subplot()
ax.scatter(x=X_base, y=Y_base, c='r', s=25)
plt.xlim(0, 8)
plt.ylim(0, 8)
plt.xlabel('x')
plt.ylabel('y')
plt.show()
```

```python
pca_1 = PCA(n_components=1)
U_axis = pca_1.fit_transform(X=np.vstack((X_base, Y_base)).T)

plt.plot(U_axis, np.zeros_like(U_axis), 'x')
plt.show()
```

```python
# Principal Component 1
# U = first_component * x + second_component * y
pca_1.components_
```

```python
pca_2 = PCA(n_components=2)
U_V_axis = pca_2.fit_transform(X=np.vstack((X_base, Y_base)).T)
```

```python
plt.plot(U_V_axis[:, 0], U_V_axis[:, 1], 'x')
plt.show()
```

```python
# Principal Component 1
# U = first_component * x + second_component * y

# Principal Component 2
# V = first_component * x + second_component * y
pca_2.components_
```
